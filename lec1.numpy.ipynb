```python
--------->>> Stacking<<<----------
```

```python
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
print(a5)
a4
```

```python
# Horizontal Stacking
np.hstack((a4,a5))
np.hstack((a5,a5,a4))
```

```python
# Vertical Stacking
print(np.vstack((a4,a5)))
np.vstack((a4,a5,a4))
```

```python
# Vertical Splitting
print(a4)
np.vsplit(a4,3)
```
