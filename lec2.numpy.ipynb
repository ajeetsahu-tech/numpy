```python
   ------> Speed <-----
```

```python
# From list
a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]
c = []
import time
start = time.time()
for i in range(len(a)):
    c.append(a[i]+b[i])
print(time .time()-start)
```

```python
# From numpy
a = np.arange(10000000)
b = np.arange(10000000,20000000)
start = time.time()
c = a + b
print(time.time()- start)
```

```python
2. --------->> Boolean Indexing <<--------
```

```python
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
# Find all numbers greater than 50.
print(a>50)   # Ye sari value jo 50 se badi hogi use True likh kar dega
a[a>50]   #<------ This is the answer.
```

```python
# Find out even numbers
a[a%2 == 0]
```

```python
# Find all number greater than 50 and are even
a[(a > 50) & (a%2 == 0)]
```

```python
#Find all numbers not divisible by 7
a[a%7 != 0]
```

```python
   ---------->>>> Broadcasting <<<<----------
```

```python
# Same shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print(a)
print(b)
print(a+b)
```

```python
# Different shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(b)
print(a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = 5
print(a)
print(a+b)
```

```python
print(np.ones((3,3)) + np.arange(3))
print(np.arange(3).reshape(3,1) + np.arange(3))
```

```python
# More Example
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)
print(a+b)             # it could not be broadcast beacuase rule can not be apply.
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)
print(a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)
print(a+b)
```

```python
a = np.array([1])
# shape -----> (1,1)
b = np.arange(4).reshape(2,2)
# shape -----> (2,2)
print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(12).reshape(4,3) 
print(a+b)                         # could not be broadcast because both are not shape same.
```

```python
a = np.arange(16).resahpe(4,4)
b = np.arange(4).reshape(2,2)
print(a+b)           # confuse nhi hona hai ye bhi same nhi hai
```

```python
   -------->> Sigmoid <<---------
```

```python
def sigmoid(array):
    return 1/(1 + np.exp(-(array)))

a = np.arange(100)
sigmoid(a)
```

```python
------->> Mean Square Error <<--------
```

```python
actual = np.random.randint(1,100,25)
predicted = np.random.randint(1,100,25)

def mse(actual, predicted):
   return np.mean((actual - predicted)**2)

print(actual)
print(predicted)
print(mse(actual, predicted))
```
