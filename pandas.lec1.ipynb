```python
import numpy as np
import pandas as pd
```

```python
  ------>>> Series From Lists <<<--------
```

```python
# String
country = ['India', 'Pakistan', 'Nepal', 'Srilanka']
pd.Series(country)
```

```python
# Integer
runs = [55, 34, 89,123, 5]
pd.Series(runs)
```

```python
# Custom index
marks = [78, 92, 93, 82]
subjects = ['math', 'english', 'hindi', 'science']
pd.Series(marks, index = subjects)
```

```python
# setting a name
marks = pd.Series(marks,index = subjects, name = 'Nitish ke Marks')
marks
```

```python
       ------->> Series From Dict <<-------
```

```python
marks = {
    'marks' : 67,
    'english' : 57,
    'science' : 89,
    'hindi' : 100
}
marks_series = pd.Series(marks, name = 'Ajeet ke marks')
marks_series
```

```python
---------->>>> Series Attribute <<<<----------
```

```python
# size
marks_series.size
```

```python
# dtype
print(marks_series.dtype)
```

```python
# name
marks_series.name
```

```python
# is_unique
marks_series.is_unique
```

```python
# index
marks_series.index
```

```python
# values
marks_series.values
```

```python
cars = pd.read_csv("cars.csv")
cars
```

```python
---------->>>> Series Method <<<<---------
```

```python
# head and tail
```

```python
cars.head(2)
```

```python
cars.tail()
```

```python
# sample
```

```python
cars.sample(3)
```

```python
# value_counts
```

```python
cars.value_counts()
```

```python
# sort_value
cars.sort_value # ------>> (ye run nhi hoga kyuki cars_csv file iske liye sahi nhi hai.)
cars.sort_value(ascending = False) # -------->> isse value lowest se highest sort hogi.
```

```python
cars.sort_value(inplace = True) # ( ye original value me changes kar deta hai. So be careful.)
```

```python
# sort_index
cars.sort_index()
cars.sort_index(inplace = True)
```
